### Controlando las excepciones con `try-except`

Las excepciones pueden ser capturadas y manejadas de forma adecuada, sin afectar el funcionamiento del programa, para ello debemos usar un `try-except`.

La estructurta de un try-except es la siguiente:

```python
try:
    <bloque de código a ejecutar>
except <excepción nombrada>:
    <acción a ejecutar>
```

`try-except` funciona de la siguiente manera

- Se ejecuta el bloque de código de `try`
- Si no ocurre ninguna excepción, el bloque `except` no se ejecuta y continua con el resto de código
- Si se presenta una excepción durante la ejecución en el bloque `try`, salta al bloque `except` y se verifica que la excepción lanzada coincida con la excepción nombrada, de ser así se ejecuta el bloque `except` y continua con el resto de código
 - Si la excepción lanzada no coincide con la nombrada y no existe otro `try-except` que la maneje, la excepción es lanzada y se frena la ejecución del programa.

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

 Veamos esto en la practica, primero veamos como sería el programa sin el uso del `try-except`:

````python
# El usuario ingresa un número
print("Por favor ingrese un número:")

# El valor ingresado se convierte a float
# Para que ocurra un error basta con ingresar una cadena de texto
x = float(input())
````

Como vemos el usuario no ha ingresado un número, sino una cadena de caracteres, la cual al tratar de ser convertida a `float` se lanza la excepción  <span style="color:red">*ValueError*</span>. 

Ahora como ya sabemos que excepción se lanza si ocurre esto, implementemos un `try-except`para controlarlo:

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese un número:")
    
    # El valor ingresado se convierte a float
    x = float(input())
except ValueError:
    print("El valor ingresado no corresponde a un número")
````

En este caso a pesar de que el usuario ingreso un valor no númerico el `try-except` capturo la excepción y evita que el programa se cierre abruptamente.

#### Exepción genérica

Si no se conoce la excepción que se puede generar, se puede usar la excepción generica `Exception`, pero además se puede identificar la excepción generada:

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese un número:")
    
    # El valor ingresado se convierte a float
    x = float(input())
except Exception:
    print("Se ha presentado una excepción")
````

Ahora identifiquemos la excepción:

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese un número:")
    
    # El valor ingresado se convierte a float
    x = float(input())
except Exception as ex:
    print("Se ha presentado una excepción", type(ex))
````

#### Uso de multiples `except`

Por otra parte la sentencia `try` puede contener multiples `except` para contener varias excepciones que se puedan presentar, o dentro del bloque `except` se pueden nombrar distintas excepciones usando parentesis `except(<excepcion>, <excepcion>, <excepcion>)`.

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Veamos un ejemplo, para este caso haremos una calculadora para realizar divisiones, veamos como sería sin usar `try-except`:

````python
# El usuario ingresa un número
print("Por favor ingrese el dividendio:")

# El valor ingresado se convierte a float
x = float(input())

# El usuario ingresa un número
# Para generar el error ingresaremos el cero
print("Por favor ingrese el divisor:")

# El valor ingresado se convierte a float
y = float(input())

# Se realiza la operación
z = x/y

# Se imprime el resultado
print(f"El resultado de dividir {x} entre {y} es {z}")
````

Para este caso se pueden presentar varias excepciones, la primera ya antes vista <span style="color:red">*ValueError*</span> que se genera debido a que no se puede convertir el valor ingresado en `float`, la segunda es <span style="color:red">*ZeroDivisionError*</span> lanzada al realizar una división entre cero.

Ahora veamos como las podemos controlar:

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese el dividendio:")

    # El valor ingresado se convierte a float
    x = float(input())

    # El usuario ingresa un número
    print("Por favor ingrese el divisor:")

    # El valor ingresado se convierte a float
    y = float(input())

    # Se realiza la operación
    z = x/y

    # Se imprime el resultado
    print(f"El resultado de dividir {x} entre {y} es {z}")
except ValueError:
    print("El valor ingresado no corresponde a un número")
    
except ZeroDivisionError:
    print("No se puede dividir el valor por cero")
````

O se puede realizar tambien de la siguiente manera:

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese el dividendio:")

    # El valor ingresado se convierte a float
    x = float(input())

    # El usuario ingresa un número
    print("Por favor ingrese el divisor:")

    # El valor ingresado se convierte a float
    y = float(input())

    # Se realiza la operación
    z = x/y

    # Se imprime el resultado
    print(f"El resultado de dividir {x} entre {y} es {z}")
except (ValueError, ZeroDivisionError):
    print("Por favor revise que los valores ingresados correspondan a un número o que el divisor sea diferente de cero")
````

#### Uso de `else` con `try-except`

La sentencia `else` puede ser usada dentro del bloque `try-except`, y solo se ejecutara si no ha ocurrido ninguna excepción.

```python
try:
    <bloque de código a ejecutar>
except <excepción nombrada>:
    <acción a ejecutar>
else:
    <acción a ejecutar>
```

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese el dividendio:")

    # El valor ingresado se convierte a float
    x = float(input())

    # El usuario ingresa un número
    print("Por favor ingrese el divisor:")

    # El valor ingresado se convierte a float
    y = float(input())

    # Se realiza la operación
    z = x/y

    #Se imprime el resultado
    print(f"El resultado de dividir {x} entre {y} es {z}")
except (ValueError, ZeroDivisionError):
    print("Por favor revise que los valores ingresados correspondan a un número o que el dividendo sea diferente de cero")
else:
    print("La calculadora ha funcionado correctamente")
````

#### Uso de `finally` con `try-except`

La sentencia `finally` se añade al final del bloque y se ejcuta siempre, es decir, se haya presentado o no una excepción.

```python
try:
    <bloque de código a ejecutar>
except <excepción nombrada>:
    <acción a ejecutar>
else:
    <acción a ejecutar>
finally:
    <acción a ejecutar>
```

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

````python
try:
    # El usuario ingresa un número
    print("Por favor ingrese el dividendio:")

    # El valor ingresado se convierte a float
    x = float(input())

    # El usuario ingresa un número
    print("Por favor ingrese el divisor:")

    # El valor ingresado se convierte a float
    y = float(input())

    # Se realiza la operación
    z = x/y

    # Se imprime el resultado
    print(f"El resultado de dividir {x} entre {y} es {z}")
except (ValueError, ZeroDivisionError):
    print("Por favor revise que los valores ingresados correspondan a un número o que el dividendo sea diferente de cero")
else:
    print("La calculadora ha funcionado correctamente")
finally:
    print("El bloque de control ha terminado su ejecución")
````
