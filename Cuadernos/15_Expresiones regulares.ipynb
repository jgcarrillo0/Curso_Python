# Capítulo 15: Expresiones regulares

Los métodos vistos en el capítulo anterior son muy útiles al momento de manipular cadenas, pero en ocasiones no son suficientes para manejar cadenas de forma óptima, por ejemplo, en la búsqueda de patrones.

Para ello se hace uso de las expresiones regulares o también conocidas como regex, con ellas podemos buscar subcadenas o patones dentro de otras cadenas.

<a class="anchor" id="sec1"></a>
## Expresiones regulares

<a class="anchor" id="subsec1"></a>
### Definición

Las expresiones regulares son secuencias de caracteres que forman un patrón de búsqueda, este patrón se aplica sobre un texto para producir un subconjunto o una versión modificada del texto original.

Las expresiones regulares pueden contener patrones de coincidencia literal, de repetición, composición, ramificación, entre otros.

<a class="anchor" id="subsec2"></a>
### Componentes de las expresiones regulares

Los componentes de la sintaxis de las expresiones regulares son:

**Literales:** Cualquier caracter se encuentra a si mismo, a menos que sea un metacaracter. 

**Secuencias de escape:** 

| Secuencia de escape | Significado    |
|---------------------|----------------|
|          \n         | Salto de línea |
|          \t         | Tabulador      |
|          \\\         | Barra diagonal |

**Conjunto de caracteres:** Al especificar caracteres entre `[]`, se buscara cualquiera de los caracteres que se encuentren en medio de ellos.

**Metacaracteres** 

Son caracteres especiales que ayudan a definir las condiciones de búsqueda, a partir de ellos podemos construir patrones avanzados y muy específicos.

Podemos agrupar los metacaracteres dependiendo de la función que cumplen, veamos los más usados:

**Delimitadores**

Permiten delimitar la busqueda.

| **Metacaracter** | **Descripción**                |
|------------------|--------------------------------|
|         ^        | Inicio de línea                |
|         $        | Fin de línea                   |
|        \A        | Inicio de texto                |
|        \Z        | Fin de texto                   |
|         .        | Cualquier caracter en la línea |
|        \b        | Encuentra límite de la palabra |
|        \B        | Encuentra distinto a límite    |

**Clases predefinidas**

Permiten buscar un conjunto de caracteres especifico.

| **Metacaracter** | **Descripción**                        |
|------------------|----------------------------------------|
|        \w        | Un caracter alfanumérico(incluye "_")  |
|        \W        | Un caracter no alfanumérico            |
|        \d        | Un caracter numérico                   |
|        \D        | Un caracter no numérico                |
|        \s        | Cualquier espacio                      |
|        \S        | Un no espacio                          |

**Iteradores**

Cualquier elemento puede estar seguido por otro tipo de metacaracteres, los iteradores definen el número de ocurrencias del caracter previo, de un metacaracter o de una subexpresión.

| **Metacaracter** | **Descripción**              |
|------------------|------------------------------|
|         *        | Cero o más coincidencias     |
|         +        | Una o más coincidencias      |
|         ?        | Cero o una coincidencia      |
|        {n}       | Exactamente n coincidencias  |
|       {n,}       | Por lo menos n coincidencias |
|       {n,m}      | Entre n y m coincidencias    |

**Subexpresiones**

Se pueden extraer subexpresiones de una expresión regular indicando por medio de `()` la parte en concreto a extraer.

**búsquedas múltiples**

Se pueden especificar varias alternativas de búsqueda usando `|`, es decir, `primera|segunda|tercera`.

### Módulo RE de Python

Una vez hemos visto una pequeña introducción de lo que son las expresiones regulares, veamos como las podemos aplicar.

El módulo `re` en Python es un módulo de la biblioteca estándar de Python, proporciona herramientas para trabajar con expresiones regulares, que son una forma de buscar y manipular cadenas de texto basándose en patrones. Las expresiones regulares son muy poderosas para tareas como la búsqueda y el reemplazo de texto, la validación de datos y la extracción de información de cadenas de texto.

Para hacer uso de este módulo debemos importarlo.

```python
# Importamos el módulo
import re
```

El módulo `re` cuenta con algunos métodos ya incorporados:

<a class="anchor" id="subsec4"></a>
### Métodos de coincidencia

#### `search(patrón, cadena, banderas=0)`

Permite buscar un patrón en un cadena. Retorna un objeto tipo `Match` de la primera coincidencia, de lo contrario retorna un `None`. 

```python
# Definimos un texto
texto = "Lo bueno de la ciencia es que es cierta independientemente de si crees o no en ella"

# Aplicamos el método, patron a buscar: ciencia
re.search("ciencia", texto)
```

Podemos usar el objeto `Match` para realizar un bloque condicional con él.

```python
# Aplicamos el método
palabra = "ciencia"
encontrado = re.search(palabra, texto)

if encontrado:
    print("Se ha encontrado la palabra:", palabra)
else:
    print("No se ha encontrado la palabra:", palabra)
```

Además, el objeto `Match` retornado brinda algunas opciones:

```python
# Posición donde empieza la coincidencia
print(encontrado.start()) 
# Posición donde termina la coincidencia
print(encontrado.end())  
# Tupla con posiciones donde empieza y termina la coincidencia
print(encontrado.span())   
# Cadena sobre la que se ha realizado la búsqueda
print(encontrado.string) 
# Cadena de coincidencia
print(encontrado.group()) 
```

#### `match(patrón, cadena, banderas=0)`

Este método es similar a `search()`, solo que busca al principio de la cadena.

```python
# Definimos un texto
texto = "hola mundo"

# Aplicamos el método
re.match("hola", texto)
```

#### `findall(patrón, cadena, banderas=0)`

Este método retorna una lista con todas las coincidencias halladas.

```python
# Definimos un texto
texto = "bar foo bar foo barbarfoo foofoo foo bar"

# Aplicamos el método
re.findall("foo", texto)
```

#### `finditer(patrón, cadena, banderas=0)`

Es un método similar a `findall()`, la diferencia es que retorna un iterador.

```python
# Definimos un texto
texto = "bar foo bar foo barbarfoo foofoo foo bar"

# Aplicamos el método
iterador = re.finditer("foo", texto)
```

```python
#Iteramos sobre el iterador
for i in iterador:
    print(i)
```

### Métodos de modificación

#### `split(patrón, cadena, maxsplit=0, banderas=0)`

Divide una cadena en cada lugar donde encuentre una coincidencia con el patrón indicado, este método retorna una lista.

```python
# Definimos un texto
texto ='''No es oro todo lo que reluce,
ni toda la gente errante anda perdida;
a las raíces profundas no llega la escarcha;
el viejo vigoroso no se marchita.
De las cenizas subirá un fuego,
y una luz asomará en las sombras;
el descoronado será de nuevo rey,
forjarán otra vez la espada rota.
'''
# Aplicamos el método
# Patron: uno o mas caracteres no alfanumericos
lista = re.split("\W+", texto)

# Mostramos la lista
lista
```

Podemos cambiar el patrón por un salto de línea:

```python
# Aplicamos el método
lista = re.split("\n", texto)

# Mostramos la lista
lista
```

#### `sub(patrón, repl, cadena, count=0, banderas=0)`

Encuentra todas las coincidencias en el texto y luego los reemplaza por un nuevo texto.

```python
# Definimos un texto
texto ='''No es oro todo lo que reluce,
ni toda la gente errante anda perdida;
a las raíces profundas no llega la escarcha;
el viejo vigoroso no se marchita.
De las cenizas subirá un fuego,
y una luz asomará en las sombras;
el descoronado será de nuevo rey,
forjarán otra vez la espada rota.
'''
# Aplicamos el método
texto_nuevo = re.sub("o", "u", texto)

# Imprimimos el texto
print(texto_nuevo)
```

#### `subn(patrón, repl, cadena, recuento=0, banderas=0)`

Este método es similar a `sub()`, pero además de devolver el texto modificado, retorna el número de cambios realizados en una tupla.

```python
# Definimos un texto
texto ='''No es oro todo lo que reluce,
ni toda la gente errante anda perdida;
a las raíces profundas no llega la escarcha;
el viejo vigoroso no se marchita.
De las cenizas subirá un fuego,
y una luz asomará en las sombras;
el descoronado será de nuevo rey,
forjarán otra vez la espada rota.
'''
# Aplicamos el método
texto_nuevo = re.subn("o", "i", texto)

# Imprimimos el texto
print(texto_nuevo)
```

<a class="anchor" id="subsec6"></a>
### Método de compilación

#### `compile(patrón, banderas=0)`

Este metodo permite compilar el patrón de busqueda, esto optimiza la busqueda, además que podemos añadirle banderas de busqueda.

```python
# Compilamos el patrón de busqueda
# Patron: Busca palabras finalizadas en o y que esten antecedidas
# por uno o mas caracteres alfanumerico
patron = re.compile(r'\w{1,}o\b')

# Definimos un texto
texto ='''No es oro todo lo que reluce,
ni toda la gente errante anda perdida;
a las raíces profundas no llega la escarcha;
el viejo vigoroso no se marchita.
De las cenizas subirá un fuego,
y una luz asomará en las sombras;
el descoronado será de nuevo rey,
forjarán otra vez la espada rota.
'''
# Aplicamos los métodos
print("##########################\nMétodo search:")
print(patron.search(texto))
print("##########################\nMétodo match:")
print(patron.match(texto))
print("##########################\nMétodo findall:")
print(patron.findall(texto))
print("##########################\nMétodo finditer.")
print(patron.finditer(texto))
print("##########################\nMétodo split.")
print(patron.split(texto))
print("##########################\nMétodo sub.")
print(patron.sub("hola", texto))
print("##########################\nMétodo subn:")
print(patron.subn("hola", texto))
```

<a class="anchor" id="subsec7"></a>
### Banderas de compilación

Las banderas de compilación permiten modificar el funcionamiento de las expresiones regulares. Podemos aplicar varias banderas usando el operador `|`.

| **Bandera** | **Forma abreviada** | **Descripción**                                                                                          |
|-------------|---------------------|----------------------------------------------------------------------------------------------------------|
|  Ignorecase |          I          | No hay diferencia entre mayúsculas o minúsculas                                                          |
|   Verbose   |          X          | Permite organizar el patrón de forma sencilla                                                            |
|    Asccii   |          A          | Las secuencias de escape `\w`, `\b`, `\s`, `\d` funcionen para coincidencias con caracteres ASCII        |
|    Dotall   |          S          | Permite que el metacaracter funcione para cualquier caracter                                             |
|    Locale   |          L          | Permite que `\w`, `\W`, `\b`, `\B`, `\s`, `\S` dependan de la ubicación actual                           |
|  Multiline  |          M          | Habilita la coincidencia para múltiples líneas, afecta el funcionamiento de los metacaracteres `^` y `$` |

```python
# Bandera Ignorecase
# Define la cadena
text = '''Por tus ojos verdes yo me perdería,
sirena de aquellas que Ulises, sagaz,
amaba y temía.
Por tus OJOS verdes yo me perdería.
Por tus oJoS verdes en lo que, fugaz,
brillar suele, a veces, la melancolía;
por tus ojos verdes tan llenos de paz,
misteriosos como la esperanza mía;
por tus OJOs verdes, conjuro eficaz,
yo me salvaría.
'''
# Compilamos el patron
patron = re.compile(r'ojos\b', re.I)

# Cambiamos ojos por labios
nuevo = patron.sub("labios", text)

print(nuevo)
```

```python
# BanderaVerbose
# Define el texto
correos = """raul.lopez@relopezbriega.com, Raul Lopez Briega,
foo bar, relopezbriega@relopezbriega.com.ar, raul@github.io, 
https://relopezbriega.com.ar, https://relopezbriega.github.io, 
python@python, river@riverplate.com.ar, pythonAR@python.pythonAR
"""
# Compilamos el patron
correo = re.compile(r"""
\b             # comienzo de delimitador de palabra
[\w.%+-]       # usuario: Cualquier caracter alfanumerico mas los signos (.%+-)
+@             # seguido de @
[\w.-]         # dominio: Cualquier caracter alfanumerico mas los signos (.-)
+\.            # seguido de .
[a-zA-Z]{2,6}  # dominio de alto nivel: 2 a 6 letras en minúsculas o mayúsculas.
\b             # fin de delimitador de palabra
""", re.X)

# Retorna una lista con los correos que cumplan la estructura
correo.findall(correos)
```

**Nota:** Para practicar de forma interactiva con las expresiones regulares puede visitar [regex](https://regex101.com/)

```python
# Importamos el módulo
import re

# Definimos la función que encuentre direcciones de correo electronico
def Buscador_correos(archivo):
    #Crea una lista vacia
    lista = []
    #Define el patron
    patron = re.compile(r"""
    \b             # comienzo de delimitador de palabra
    [\w.%+-]       # usuario: Cualquier caracter alfanumerico mas los signos (.%+-)
    +@             # seguido de @
    [\w.-]         # dominio: Cualquier caracter alfanumerico mas los signos (.-)
    +\.            # seguido de .
    [a-zA-Z]{2,6}  # dominio de alto nivel: 2 a 6 letras en minúsculas o mayúsculas.
    \b             # fin de delimitador de palabra
    """, re.X)
    # Abre el archivo
    archivo = open(archivo)
    # Itera sobre el archivo
    for line in archivo:
        line = line.rstrip()
        Correos_encontrado = re.findall(patron ,line)
        lista += Correos_encontrado
        
    # Dejamos solo correos unicos    
    unicos = set(lista)
    return unicos
    
# Llamdo de la funcion
Buscador_correos("recursos\data\mbox-short.txt")
    
```
