# Capítulo 12: Estructuras de control: While

En ocaciones necesitaremos ejecutar un bloque de código hasta que se cumpla cierta condicón, para realizar esto Python nos ofrece el bucle `while`.

<a class="anchor" id="sec1"></a>
## Bucle While

<a class="anchor" id="subsec1"></a>
### Definición

`while` es un tipo de bucle que permite ejecutar un bloque de código muchas veces, para ello el usuario debe definir una condición y mientras esta se cumpla el bucle se ejecutará, de lo contrario finalizará el bucle. Se debe tener en cuenta que la condicón definida debe estar bien parametrizada, pues de lo contrario el bucle se puede quedar "infinaitamente" ejecutando el mismo bloque.

La estructura de un bucle `while` es:

```python
while <condición>:
    <acción a ejecutar>
```

<a class="anchor" id="subsec2"></a>
### Uso del bucle `while`

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Supongamos que queremos realizar una suma hasta que alcance cierto valor:

````python
# Define la variable
x = 0

# Define el bucle while
while x < 40:
    x += 2
    print(x)
````

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Ahora iteremos sobre una cadena:

````python
# Define la cadena
text = "Interestellar"

# Define la variable
i = 0

# Define el bucle while
while i < len(text):
    print(text[:i + 1]) 
    i += 1
````

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Generemos un arbolito de navidad con `while`:

````python
# Define variables
z = 7
x = 1

# Define el bucle while
while z > 0:
    print(' ' * z + '▲' * x + ' ' * z)
    x += 2
    z -= 1
````

<a class="anchor" id="subsec3"></a>
### Uso de `while` anidados

Los bucles `while` tambien se pueden anidar, veamos un ejemplo para generar todas las conmbinaciones de dos numeros hasta 2.

````python
i = 0
j = 0
while i < 3:
    while j < 3:
        print(i,j)
        j += 1
    i += 1
    j = 0
````

<a class="anchor" id="subsec4"></a>
### Uso de la sentencia `continue` y `break`

De igual manera que con el bucle `for`, `while` tambien permite usar las sentencias `continue` y `break`.

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Busquemos todos los multiplos de 11 entre el  0 y el 100:

````python
# Define la variable
x = -1

# Define el bucle
while x < 100:
    x += 1
    if x%11 != 0:
        continue
    print (x)
````

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

Busquemos todos los multiplos de 11 hasta que supere un valor ingresado:

````python
# Define la variable
y = int(input())
````

````python
# Define el bucle
x = -1
while True:
    x += 1
    if x%11 != 0:
        continue
    print (x)
    if x > y:
        break
````
