# Capítulo 14: Manejo de cadenas

En Python, una cadena es una secuencia de caracteres inmutable.

<a class="anchor" id="sec1"></a>
## Cadena de caracteres

<a class="anchor" id="subsec1"></a>
### Definición

Como ya antes lo habíamos mencionado las cadenas de caracteres son una secuencia de caracteres delimitados por `"` o `'`. En Python podemos manipular y transformar las cadenas de una forma sencilla a partir del uso de métodos. 

Saber manipular o transformar cadenas de caracteres es de gran importancia en el campo del análisis y procesamiento de texto, dado que permite extraer información valiosa; por ejemplo, podemos usar los comentarios de Facebook y realizar un análisis de sentimientos.

<a class="anchor" id="subsec2"></a>
### Acceso a un caracter de una cadena

Recordemos que las cadenas de caracteres representan objeto iterable ordenado, eso significa que cada uno de los caracteres está asociado a un índice, el cual nos indica la posición en la que se encuentra:

<p style="text-align:center;">
  <img src="recursos/imagenes/cadena.png" width="60%"/>
</p>

Lo anterior permite seleccionar partes específicas de la cadena haciendo uso de la notación `[inicio:fin:salto]`.

<div class = "alert alert-block alert-warning">
    <b>📝 Ejemplo</b>
</div>

```python
# Definimos la cadena
texto = "mi primera cadena"

# Seleccionamos desde la posición 3 hasta la 11
seleccion = texto[3:11]
print(seleccion)
```

Algo que se suele ver, es omitir los valores de inicio o final; en el caso de `[:fin]` se da a entender que se toma desde la posición inicial y `[inicio:]` toma hasta la posición final.

```python
# Seleccionamos desde la posición 0 hasta la 2
seleccion = texto[:2]
print(seleccion)
```

<a class="anchor" id="subsec3"></a>
### Concatenar cadenas

Por otra parte podemos concatenar cadenas con el operador `+`, esto es útil para unir varias cadenas:

```python
# Definimos las cadenas
cad = " "
cad1 = "primera"
cad2 = "segunda"
cad3 = "tercera"

# Concatenamos las cadenas
cad4 = cad1 + cad + cad2 + cad + cad3
print(cad4)
```

<a class="anchor" id="subsec4"></a>
### Repetir cadenas

También podemos usar el operador `*` para repetir una cadena cuantas veces deseemos:

```python
# Definimos la cadena
cad = "Python "

# Repitamos la cadena 8 veces
cadmult = cad * 8
print(cadmult )
```

### Métodos

#### `count()`

Este método permite contar el número de veces que se encuentra un conjunto de caracteres.

```python
# Definimos la cadena
cad = "No hay nada que un hombre no sea capaz de hacer cuando una mujer le mira"

# Cuenta el número de veces que está la vocal a
cad.count("a")
```

#### `find()`

El método `find()` retorna el índice donde se encuentra la primera coincidencia del conjunto que se pase como argumento. Si la cadena que se pasa como argumento no se encuentra, retorna un -1. 

Si se desea buscar la última coincidencia del conjunto que se pasa como argumento use `rfind()`.

```python
# Busca donde se encuentra la palabra hombre
cad.find("hombre")
```

```python
# Busca la ultima aparición de la vocal a
cad.rfind("a")
```

<a class="anchor" id="subsec6"></a>
### Métodos de verificación

#### `startswith()` y `endswith()`

Estos dos métodos retornan un valor booleano si la cadena empieza o termina con el conjunto de caracteres que se pase como argumento.

```python
# Definimos la cadena
cad = "soy una frase corta"

# Se realiza la validación
empiezacon = cad.startswith("soy")
terminacon = cad.endswith("corta")

print("¿La cadena empieza con la palabra, soy?", empiezacon)
print("¿La cadena termina con la palabra, corta?", terminacon)
```

#### `isalnum()`

Determina si todos los caracteres son alfanuméricos.

```python
# Definimos la cadena
cad = "abcd123"

# Aplicamos el método
cad.isalnum()
```

#### `isalpha()`

Determina si todos los caracteres son alfabéticos.

```python
# Definimos la cadena
cad = "abcdefgh"

# Aplicamos el método
cad.isalpha()
```

#### `islower()`

Determina si todas las letras son minúsculas.

```python
#Definimos la cadena
cad = "abcdefgh"

#Aplicamos el método
cad.islower()
```

#### `isupper()`

Determina si todas las letras son mayúsculas.

```python
# Definimos la cadena
cad = "ABCDEFG"

# Aplicamos el método
cad.isupper()
```

#### `isprintable()`

Determina si la cadena contiene todos los caracteres son imprimibles.

```python
# Definimos la cadena
cad = "Hola \n mundo"

# Aplicamos el método
cad.isprintable()
```

#### `isspace()`

Determina si la cadena contiene solo espacios.

```python
# Definimos la cadena
cad = "         "

# Aplicamos el método
cad.isspace()
```

#### `isdigit()`

Determina si la cadena representa números.

```python
# Definimos la cadena
cad = "\u00b3" #Representa ³, superindice 3

# Aplicamos el método
cad.isdigit()
```

```python
# Definimos la cadena
cad = "\u2175" #Representa ⅕

# Aplicamos el método
cad.isdigit()
```

```python
# Definimos la cadena
cad = "\u325b" #Representa ㉛

# Aplicamos el método
cad.isdigit()
```

#### `isnumeric()`

Determina si la cadena representa números, incluye también caracteres de connotación numérica que no necesariamente son dígitos (por ejemplo, una fracción)

```python
# Definimos la cadena
cad = "\u2175" #Representa ⅕

# Aplicamos el método
cad.isnumeric()
```

```python
# Definimos la cadena
cad = "\u325b" #Representa ㉛

# Aplicamos el método
cad.isnumeric()
```

```python
# Definimos la cadena
cad = "\u00b3" #Representa ³, superindice 3

# Aplicamos el método
cad.isnumeric()
```

#### `isdecimal()`

Determina si la cadena cuenta unicamente con números digitales, es decir formados por los digitos de 0 al 9

```python
# Definimos la cadena
cad = "\u0030" #Representa 0

# Aplicamos el método
cad.isdecimal()
```

```python
# Definimos la cadena
cad = "\u2175" #Representa ⅕

# Aplicamos el método
cad.isdecimal()
```

```python
# Definimos la cadena
cad = "\u325b" #Representa ㉛

# Aplicamos el método
cad.isdecimal()
```

```python
# Definimos la cadena
cad = "\u00b3" #Representa ³, superindice 3

# Aplicamos el método
cad.isdecimal()
```

### Métodos de transformación

#### `capitalize()`

Retorna una cadena con la primera letra en mayúscula.

```python
# Definimos la cadena
cad = "titulo del libro" 

# Aplicamos el método
cad.capitalize()
```

#### `lower()`

Retornan la cadena con todas las letras en minúsculas.

```python
# Definimos la cadena
cad = "AbCdEfGhIj" 

# Aplicamos el método
cad.lower()
```

#### `upper()`

Retornan la cadena con todas las letras en mayúscula.

```python
# Definimos la cadena
cad = "AbCdEfGhIj" 

# Aplicamos el método
cad.upper()
```

#### `swapcase()`

Retornan la cadena cambiando las minúsculas por mayúsculas y viceversa.

```python
# Definimos la cadena
cad = "AbCdEfGhIj" 

# Aplicamos el método
cad.swapcase()
```

#### `strip()`, `rstrip()`, `lstrip()`

Remueven los espacios en blanco de una cadena.

```python
# Definimos la cadena
cad = "  cadena   " 
```

```python
# Aplicamos el método
# Remueve los espacios al inicio y fin de la cadena
cad.strip()
```

```python
# Aplicamos el método
# Remueve los espacios de la derecha
cad.rstrip()
```

```python
# Aplicamos el método
# Remueve los espacios de la izquierda
cad.lstrip()
```

#### `replace()`

Reemplaza una cadena por otra.

```python
# Definimos la cadena
cad = "Clase de español" 

# Aplicamos el método
cad.replace("español", "programación")
```

### Métodas de separación y Unión

#### `split()`

El método `split()` permite dividir una cadena según un carácter separador, por defecto el método usa espacio en blanco; al aplicar el método se retorna una lista con la cadena separada.

```python
# Definimos una cadena
cad = "esto es una cadena que queremos separar"

# Aplicamos el método, usa los espacios por defecto
cad.split()
```

Se puede pasar un segundo argumento para indicar el maximo de divisiones.

```python
# Definimos una cadena
cad = "esto es una cadena que queremos separar"

# Aplicamos el método, usa los espacios por defecto
cad.split(" ", 3)
```

#### `join()`

Permite unir una lista de caracteres, para poderlo aplicar, debemos invocarlo sobre una cadena en la cual se indica el separador. Note que es el proceso opuesto al método `split()`.

```python
# Definamos una lista
lista = ['esto', 'es', 'una', 'cadena', 'que', 'queremos', 'separar']

# Aplicamos el método
" ".join(lista)
```
